#Reading data from CSV_FILE_Path.py and reading all the files and writing it delta lake

```python
from CSV_File_Path import INGESTION_CONFIG
```

```python
for item in INGESTION_CONFIG:
    print(f"Importing file path {item['path']} and writing to table bronze.{item['table']}")

    df = (
        spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .csv(item["path"])
    )

    (
        df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(f"bronze.{item['table']}")
    )
```
